## 1. Mathematical background

A pure single-qubit state can be written as

$$|\\psi\\rangle = \\alpha|0\\rangle + \\beta|1\\rangle = \\begin{pmatrix}\\alpha\\\\\\beta\\end{pmatrix},$$

where $\\alpha$ and $\\beta$ are complex probability amplitudes. A valid state is normalized:

$$\\langle\\psi|\\psi\\rangle = |\\alpha|^2 + |\\beta|^2 = 1.$$

The probabilities of measuring $0$ and $1$ in the computational basis are $|\\alpha|^2$ and $|\\beta|^2$.

## 3. Create a complex state vector

Consider the unnormalized vector $|\\phi\\rangle = (1, i\\sqrt{3})^T$. Use `dtype=complex` so NumPy stores complex numbers.

```python
phi = np.array([1, 1j * np.sqrt(3)], dtype=complex)

print("phi =", phi)
print("shape =", phi.shape)
print("dtype =", phi.dtype)
```

## 4. Compute the norm and inner product

The norm is $\\|\\phi\\| = \\sqrt{\\langle\\phi|\\phi\\rangle}$. For complex arrays, `np.vdot(a, b)` conjugates the first argument, so it is suitable for the inner product $a^\\dagger b$.

```python
inner_product = np.vdot(phi, phi)
norm = np.linalg.norm(phi)

print("<phi|phi> =", inner_product)
print("||phi|| =", norm)
print("Already normalized?", np.isclose(inner_product, 1.0))
```

## 5. Normalize the vector

Divide a nonzero vector by its norm. Then check the defining condition $\\langle\\psi|\\psi\\rangle=1$.

```python
psi = phi / norm

print("psi =", psi)
print("<psi|psi> =", np.vdot(psi, psi))
print("Normalized?", np.isclose(np.vdot(psi, psi), 1.0))
```

## 6. Calculate measurement probabilities

For a state $|\\psi\\rangle=(\\alpha,\\beta)^T$, the computational-basis probabilities are $P(0)=|\\alpha|^2$ and $P(1)=|\\beta|^2$. In NumPy, use `np.abs(psi) ** 2`.

```python
probabilities = np.abs(psi) ** 2

print("P(0) =", probabilities[0])
print("P(1) =", probabilities[1])
print("Total probability =", probabilities.sum())

# Numerical checks
assert np.all(probabilities >= 0)
assert np.isclose(probabilities.sum(), 1.0)
```

## 8. What to remember

- A quantum state uses complex probability amplitudes, not probabilities directly.
- A normalized state has unit norm.
- `np.vdot(a, b)` conjugates `a`; `np.dot(a, b)` does not do that for complex vectors.
- Measurement probabilities in a complete orthonormal basis are non-negative and sum to 1.

The theory explains which quantities must be computed. The numerical checks help catch implementation mistakes and floating-point issues.
